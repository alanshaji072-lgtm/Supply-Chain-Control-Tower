```python
# 6.How much inventory is available?

df_fact_inventory["Available_Qty"]=df_fact_inventory["On_Hand_Qty"]-df_fact_inventory["Reserved_Qty"]
df_fact_inventory["Available_Qty"].describe()
```

##### Some records have negative Available_Qty, meaning Reserved_Qty exceeds
On_Hand_Qty. This may represent an operational inventory situation and
requires further investigation.

```python
# 7. Which records are at stockout risk?

df_fact_inventory[df_fact_inventory["On_Hand_Qty"]<=0]
```

##### 21 inventory records had On_Hand_Qty <= 0. These include zero-stock situations as well as one negative on-hand observation. Several zero-stock records had inventory in transit, suggesting that replenishment may already have been underway. Some records also had reservations despite zero on-hand inventory, resulting in negative available inventory. These situations should be treated as inventory-risk events rather than automatically classified as data errors.

```python
# 8. Products below their reorder point

df_reorder_risk = df_fact_inventory[df_fact_inventory["On_Hand_Qty"] <= df_fact_inventory["Reorder_Point"]]

df_reorder_risk.shape
```

```python
df_reorder_risk[["Snapshot_Date","Warehouse_ID","Product_ID","On_Hand_Qty","In_Transit_Qty","Reorder_Point","Daily_Demand"]].head(10)
```

##### 851 of 1,499 inventory records (56.8%) were at or below their reorder point, indicating that a substantial proportion of inventory observations were in a replenishment-risk state. However, several of these records had inventory already in transit, so reorder-point risk should be evaluated alongside incoming inventory and demand before concluding that a shortage exists.

```python
# 9. Safety-stock risk

df_safety_risk = df_fact_inventory[df_fact_inventory["Available_Qty"] <= df_fact_inventory["Safety_Stock"]]
df_safety_risk.shape
```

```python
df_safety_risk.head(10)
```

##### 657 of 1,499 inventory records (43.8%) had available inventory at or below their safety-stock level, indicating that a significant proportion of observations were operating with limited inventory buffers. Several of these records had inventory in transit, suggesting that replenishment may already have been underway. Therefore, safety-stock risk should be interpreted together with incoming inventory and demand rather than treated as an immediate stockout.

```python
# 10. Days of inventory coverage

import numpy as np

df_fact_inventory["Inventory_Risk"]=np.select([
    df_fact_inventory["Available_Qty"]<0,
    df_fact_inventory["Daily_Demand"]==0,
    df_fact_inventory["Days_of_Cover"]==2,
    df_fact_inventory["Days_of_Cover"]==5,
    df_fact_inventory["Days_of_Cover"]==10
    ],
    [
        "Critical",
        "Not Applicable",
        "Critical",
        "Low",
        "Medium"
    ],
    default="Healthy"
    )
df_fact_inventory["Inventory_Risk"].value_counts(dropna=False)
```

```python
df_fact_inventory.head(20)
```

##### The inventory-risk classification identified 36 Critical, 28 Low, 2 Medium, 1,433 Healthy, and 1 Not Applicable observations. The majority of observations were classified as Healthy under the current Days-of-Cover thresholds. However, reorder-point and safety-stock checks identified substantially more risk observations, indicating that inventory risk depends on multiple operational indicators rather than Days of Cover alone.

##### The average transit time was 24.61 days, with a median of 25 days across 2,701 shipments. Transit times ranged from 3 to 48 days, indicating substantial variation in shipment duration. The middle 50% of shipments had transit times between 18 and 32 days.

##### Ocean was the most frequently used shipping mode, accounting for 1,601 of 2,701 shipments (59.3%), followed by Air at 673 (24.9%) and Rail at 427 (15.8%). Ocean therefore represents the largest share of shipment activity in the dataset.

```python
# 5. Port and customs delays
df_fact_shipments[["Port_Dwell_Days","Customs_Delay_Days"]].describe()
```

The maximum customs delay is substantially higher than the typical values and should be investigated as a potential exceptional event rather than treated as representative of normal customs performance.
